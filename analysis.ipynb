```python
# Load the CSV files
path_field = '/Users/spencerzhang/GitHub/PhD/North-America-Gas-2021/revision_data/field/'
path_aerial = '/Users/spencerzhang/GitHub/PhD/North-America-Gas-2021/revision_data/GHGRP equipment data/'
df_fields = pd.read_csv(path_field + 'us_field_2023.csv')
df_equipment = pd.read_csv(path_aerial + 'equipment_match_aerial_rate_lookup_table.csv')

# Preview columns to confirm merge key
print("us_field_2023 columns:", df_fields.columns)
print("equipment_match_aerial_rate_lookup_table columns:", df_equipment.columns)

# Assuming the join key is 'us_fields' in df_fields and 'Field' in df_equipment
# Update these column names if different
df_merged = df_fields.merge(df_equipment, left_on='FIPS', right_on='FIPS', how='left')

# Display merged result or save
print(df_merged.head())
```
